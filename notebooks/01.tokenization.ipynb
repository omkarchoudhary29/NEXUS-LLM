```python
from transformers import AutoTokenizer
```

```python
model_name="distilbert/distilgpt2"
```

```python
tokenizer=AutoTokenizer.from_pretrained(model_name)
```

```python
text="I Love Machine Learning"
```

```python
print(tokenizer.__class__.__name__)
print(tokenizer.vocab_size)
```

```python
tokens=tokenizer.tokenize(text)
print(tokens)
```

```python
tokens_ids=tokenizer.encode(text,add_special_tokens=False)
print(tokens_ids)
```

```python
tokens_from_ids=tokenizer.convert_ids_to_tokens(tokens_ids)
print(tokens_from_ids)
```

```python
tokens_to_text=tokenizer.decode(tokens_ids)
print(tokens_to_text)
```

```python
print("=" * 50)
print("           NEXUS LLM")
print("       TOKENIZATION STUDIO")
print("=" * 50)

print("\nOriginal Text:")
print(text)

print("\nTokens:")
print(tokens)

print("\nToken IDs:")
print(tokens_ids)

print("\nTokens From IDs:")
print(tokens_from_ids)

print("\nTotal Token Count:")
print(len(tokens))

print("\nDecoded Text:")
print(tokens_to_text)

print("=" * 50)
```

```python
test_sentences = [
    "Hello World",
    "Artificial Intelligence",
    "I am learning Large Language Models.",
    "unbelievable",
    "MachineLearning",
    "AI, ML, and Deep Learning!",
    "123456789",
    "I love Python!!!"
]

for sentence in test_sentences:

    ids = tokenizer.encode(
        sentence,
        add_special_tokens=False
    )

    tokens = tokenizer.convert_ids_to_tokens(ids)

    print("\nText:", sentence)
    print("Tokens:", tokens)
    print("Token IDs:", ids)
    print("Token Count:", len(ids))
```

```python
sentences = [
    "I love AI.",
    "I am learning Large Language Models.",
    "Transformers are powerful models."
]

print(sentences)
```

```python
# Configure padding token

tokenizer.pad_token = tokenizer.eos_token

print("PAD token:", tokenizer.pad_token)
print("PAD token ID:", tokenizer.pad_token_id)
```

```python
# 1. Without padding
result_1 = tokenizer(
    sentences,
    padding=False,
    truncation=False
)

# 2. With padding
result_2 = tokenizer(
    sentences,
    padding=True,
    truncation=False,
    return_tensors="pt"
)

# 3. With padding and truncation
result_3 = tokenizer(
    sentences,
    padding=True,
    truncation=True,
    max_length=10,
    return_tensors="pt"
)

print("Without padding:")
print([len(ids) for ids in result_1["input_ids"]])

print("\nWith padding:")
print(result_2["input_ids"].shape)

print("\nWith padding + truncation:")
print(result_3["input_ids"].shape)
```

```python
test_texts = [
    "I love artificial intelligence.",
    "Tokenization is important for LLMs!",
    "I am learning Python, PyTorch and Transformers.",
    "123456789 987654321",
    "Hello!!! How are you???",
    "MachineLearningIsAmazing",
    "AI is transforming the world in 2026."
]

for text in test_texts:

    tokens = tokenizer.tokenize(text)
    token_ids = tokenizer.encode(
        text,
        add_special_tokens=False
    )

    print("Text:", text)
    print("Tokens:", tokens)
    print("Token count:", len(token_ids))
    print("-" * 60)
```

```python
import pandas as pd

results = []

for text in test_texts:

    word_count = len(text.split())

    token_count = len(
        tokenizer.encode(
            text,
            add_special_tokens=False
        )
    )

    results.append({
        "Text": text,
        "Word Count": word_count,
        "Token Count": token_count,
        "Tokens per Word": round(token_count / max(word_count, 1), 2)
    })

token_df = pd.DataFrame(results)

print(token_df)
```

```python
from transformers import AutoTokenizer

bert_tokenizer = AutoTokenizer.from_pretrained(
    "bert-base-uncased"
)

gpt_tokenizer = tokenizer

comparison_texts = [
    "I love artificial intelligence.",
    "Unbelievable technology is transforming industries.",
    "Tokenization is important for language models.",
    "MachineLearning and DeepLearning"
]

for text in comparison_texts:

    gpt_tokens = gpt_tokenizer.tokenize(text)
    bert_tokens = bert_tokenizer.tokenize(text)

    print("\nText:", text)

    print("\nGPT-2 tokens:")
    print(gpt_tokens)
    print("GPT-2 count:", len(gpt_tokens))

    print("\nBERT tokens:")
    print(bert_tokens)
    print("BERT count:", len(bert_tokens))

    print("=" * 70)
```
